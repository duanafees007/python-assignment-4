```python
# Question # 1: List of all subject marks.
eng = 78
urdu = 77
math = 90
pst = 85
comp = 79
marks = [eng , urdu , math , pst , comp]
totalmarks = eng + urdu + math + pst + comp
averagemarks = totalmarks / 5
print("All subject Marks:" , marks)
print("marks obtained:" , totalmarks)
print("Average Marks:" , averagemarks)
```

```python
# Question # 3(a): Store student information by using dictionary method
print("STUDENT INFORMATION")
student_data = {
    "Name" : "M Aman",
    "Age" : 14,
    "Class" : 9,
    "Marks" : 450}
print("\n","Name :" , student_data["Name"])
print("Age :" , student_data["Age"])
print("Class :" , student_data["Class"])
print("Marks :" , student_data["Marks"])
print("\n", student_data)
print("\n", student_data.values())
```

```python
# Question # 4: Calculate and print the percentages from the Dictionary
students = {
    "Ali" : [80, 85, 90, 88, 92],
    "Huda" : [75, 78, 82, 80, 85],
    "Zain" : [90, 92, 88, 85, 95],
    "Sara": [78, 83, 62, 59, 85]
}
for name, marks in students.items():
    total = sum(marks)
    percentage = (total / 500 ) * 100
    print(name, "Percentage:", percentage, "%")
```

```python
# Question # 5: Dictionary method
students = {
    "Ali" : [435],
    "Huda" : [400],
    "Zain" : [450],
    "Sara": [367]
}
print("----Display all Keys----")
print("\n", students.keys())
```

```python
print("----Display all Values----")
print("\n", students.values())
```

```python
print("----Display all Items----")
print("\n", students.items())
```

```python
print("----Dictionay Update----")
students.update({"Ali": [440] , "Huda": [425]})
print("\n", students)
```

```python
students = {
    "Ali" : [435],
    "Huda" : [400],
    "Zain" : [450],
    "Sara": [367]
}
print("----Remove Student by using Pop----")
students.pop("Ali")
print("\n", students)
```
